```python
# importamos las librerias necesarias

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
```

```python
# chequeamos la información y los nulos

df.info()
df.isna().sum()
```

No tiene nulos, ya que realizamos la limpieza en el TP anterior.

```python
# definimos las variables test y train

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)
```

```python
# utilizamos el scaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# y aplicamos Pytorch a las variables test y train

X_train_tensor = torch.tensor(X_train_scaled, dtype=torch.float32)
X_test_tensor = torch.tensor(X_test_scaled, dtype=torch.float32)

y_train_tensor = torch.tensor(y_train.values, dtype=torch.float32).view(-1, 1)
y_test_tensor = torch.tensor(y_test.values, dtype=torch.float32).view(-1, 1)
```

```python
# creamos una clase llamada RedCardiaca

class RedCardiaca(nn.Module):
    def __init__(self, input_dim, neuronas=16):
        super(RedCardiaca, self).__init__()

        self.modelo = nn.Sequential(
            nn.Linear(input_dim, neuronas),
            nn.ReLU(),
            nn.Linear(neuronas, neuronas // 2),
            nn.ReLU(),
            nn.Linear(neuronas // 2, 1),
            nn.Sigmoid()
        )

    def forward(self, x):
        return self.modelo(x)
```

Voy a trabajar PyTorch con "nn.Module". Utilizare capas densas (Linear) con función de activación ReLU, ya que nos permitira modelar relaciones no lineales entre las variables.

La capa de salida tiene una sola neurona con activación Sigmoide, dado que buscamos plantear una clasificación binaria (positivo o negativo de enfermeadad).

```python
# entrenamiento y evaluación del modelo

input_dim = X_train_tensor.shape[1]

modelo = RedCardiaca(input_dim=input_dim, neuronas=16)

criterio = nn.BCELoss()
optimizador = optim.Adam(modelo.parameters(), lr=0.001)

epochs = 300

train_losses = []
test_losses = []

for epoch in range(epochs):
    modelo.train()

    y_pred_train = modelo(X_train_tensor)
    loss_train = criterio(y_pred_train, y_train_tensor)

    optimizador.zero_grad()
    loss_train.backward()
    optimizador.step()

    modelo.eval()
    with torch.no_grad():
        y_pred_test = modelo(X_test_tensor)
        loss_test = criterio(y_pred_test, y_test_tensor)

    train_losses.append(loss_train.item())
    test_losses.append(loss_test.item())

    if (epoch + 1) % 50 == 0:
        print(f"Epoch {epoch+1}/{epochs} | Loss Train: {loss_train.item():.4f} | Loss Test: {loss_test.item():.4f}")
```

```python
# Grafico

plt.figure(figsize=(8,5))
plt.plot(train_losses, label="Error entrenamiento")
plt.plot(test_losses, label="Error testeo")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Curvas de error - Red neuronal")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Evaluación de metricas y matriz de confusión

def evaluar_modelo(modelo, X_tensor, y_real, nombre="Conjunto"):
    modelo.eval()

    with torch.no_grad():
        probabilidades = modelo(X_tensor)
        predicciones = (probabilidades >= 0.5).int().numpy().flatten()

    acc = accuracy_score(y_real, predicciones)
    precision = precision_score(y_real, predicciones)
    recall = recall_score(y_real, predicciones)
    f1 = f1_score(y_real, predicciones)
    matriz = confusion_matrix(y_real, predicciones)

    print(f"--- {nombre} ---")
    print(f"Accuracy:  {acc:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall:    {recall:.4f}")
    print(f"F1-score:  {f1:.4f}")
    print("Matriz de confusión:")
    print(matriz)

    return acc, precision, recall, f1
```

```python
metricas_train = evaluar_modelo(modelo, X_train_tensor, y_train.values, "Entrenamiento")
metricas_test = evaluar_modelo(modelo, X_test_tensor, y_test.values, "Testeo")
```

El modelo presento un buen desempeño general. Pero se observa una disminución de los valores alcanzados en el test respecto del entrenamiento. Por ejemplo, Recall bajo bastante (es el que más cambio), paso de 0.87 en entrenamiento a 0.76 en el testeo. Esto nos dice que mientras antes acerto el 87% de los casos positivos, pero con el testeo solo llego al 76% y dado que es un dataset medico ese porcentaje de acierto resulta bajo dada la importancia de dar los diagnosticos acertados.

Fuera de esto, en la matriz de confunsión se ve que el modelo logra clasificar correctamente los datos en su mayoría.

### Probamos el modelo con más capas

```python
# Desarrollo varios modelos con distintas capas para evaluar y utilizar el mejor

class RedFlexible(nn.Module):
    def __init__(self, input_dim, capas):
        super(RedFlexible, self).__init__()

        layers = []
        dim_actual = input_dim

        for neuronas in capas:
            layers.append(nn.Linear(dim_actual, neuronas))
            layers.append(nn.ReLU())
            dim_actual = neuronas

        layers.append(nn.Linear(dim_actual, 1))
        layers.append(nn.Sigmoid())

        self.modelo = nn.Sequential(*layers)

    def forward(self, x):
        return self.modelo(x)
```

```python
# Entreno los modelos

def entrenar_modelo(capas, epochs=300, lr=0.001):
    modelo = RedFlexible(input_dim=X_train_tensor.shape[1], capas=capas)

    criterio = nn.BCELoss()
    optimizador = optim.Adam(modelo.parameters(), lr=lr)

    train_losses = []
    test_losses = []

    for epoch in range(epochs):
        modelo.train()

        y_pred_train = modelo(X_train_tensor)
        loss_train = criterio(y_pred_train, y_train_tensor)

        optimizador.zero_grad()
        loss_train.backward()
        optimizador.step()

        modelo.eval()
        with torch.no_grad():
            y_pred_test = modelo(X_test_tensor)
            loss_test = criterio(y_pred_test, y_test_tensor)

        train_losses.append(loss_train.item())
        test_losses.append(loss_test.item())

    return modelo, train_losses, test_losses
```

```python
# Defino la cantidad de capas que tendrá cada uno de los modelos

arquitecturas = {
    "Modelo chico (4)": [4],
    "Modelo intermedio (16, 8)": [16, 8],
    "Modelo grande (64, 32, 16)": [64, 32, 16],
    "Modelo muy grande (128, 64, 32)": [128, 64, 32]
}

resultados = {}

for nombre, capas in arquitecturas.items():
    modelo_exp, train_loss, test_loss = entrenar_modelo(capas, epochs=300)
    resultados[nombre] = {
        "modelo": modelo_exp,
        "train_loss": train_loss,
        "test_loss": test_loss
    }
```
